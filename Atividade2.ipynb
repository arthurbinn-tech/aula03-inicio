```python
quantidade = int(input("Digite a quantidade de numeros primos"))

contador = 0
numero = 2

while contador < quantidade:

  primo = True

  if numero < 2:
    primo = False

  elif numero == 2:
    primo = True

  elif numero % 2 == 0:
    primo = False

  else:
    for divisor in range (3, numero, 2):
      if numero % divisor == 0:
        primo = False
        break

  if primo:
    print(numero, end=" ")
    contador += 1

  numero += 1
```
