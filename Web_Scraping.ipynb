```python
import urllib.request
from urllib.parse import quote
import json


cidade = input("Digite o nome da cidade: ")


url = f"https://wttr.in/{quote(cidade)}?format=j1"

try:

    with urllib.request.urlopen(url) as response:
        texto = response.read().decode('utf-8')
        dados = json.loads(texto)


    hoje = dados["weather"][0]

    minima = hoje["mintempC"]
    maxima = hoje["maxtempC"]


    print("\nPREVISÃO DO TEMPO")
    print("Cidade:", cidade)
    print("Data:", hoje["date"])
    print("Temperatura mínima:", minima, "°C")
    print("Temperatura máxima:", maxima, "°C")

except Exception as erro:
    print("Erro ao consultar a previsão:", erro)
```
