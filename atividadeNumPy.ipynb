```python
import numpy as np

dados = np.array([
    [120, 15000, 100, 125.00, 4],
    [135, 16200, 108, 120.00, 5],
    [98, 11700, 82, 119.39, 3],
    [150, 21000, 125, 140.00, 7],
    [175, 24500, 140, 140.00, 9],
    [160, 22400, 132, 140.00, 6],
    [110, 13200, 95, 120.00, 2]
])

dias = np.array([
    "Segunda", "Terça", "Quarta", "Quinta", "Sexta", "Sábado", "Domingo"
])

indicadores = np.array([
    "Pedidos", "Faturamento", "Clientes", "Ticket médio", "Reclamações"
])

print("\nParte 1 — Conhecendo os dados")
print("Quantidade de linhas", dados.shape[0])
print("Quantidade de colunas", dados.shape[1])
print("Quantidade total de valores", dados.size)
print("Número de dimensões", dados.ndim)
print("Tipo dos dados", dados.dtype)

pedidos = dados[:, 0]
faturamento = dados[:, 1]
clientes = dados[:, 2]
reclamacoes = dados[:, 4]

print("\nParte 2 — Acessando os dados")
print("2.1 Dados de sexta-feira", dados[4])
print("2.2 Faturamento de todos os dias", faturamento)
print("2.3 Reclamações de todos os dias", reclamacoes)
print("2.4 Dados de quinta-feira a domingo")
print(dados[3:])
print("2.5 Colunas de pedidos, faturamento e clientes")
print(dados[:, :3])

total_pedidos = np.add.reduce(pedidos)
total_faturamento = np.add.reduce(faturamento)
total_clientes = np.add.reduce(clientes)
media_pedidos = np.mean(pedidos)
media_faturamento = np.mean(faturamento)
media_reclamacoes = np.mean(reclamacoes)

print("\nParte 3 — Operações numéricas")
print(f"3.1 Total de pedidos — {total_pedidos:.0f}")
print(f"3.2 Faturamento total — R$ {total_faturamento:.2f}")
print(f"3.3 Total de clientes — {total_clientes:.0f}")
print(f"3.4 Média diária de pedidos — {media_pedidos:.2f}")
print(f"3.5 Média diária de faturamento — R$ {media_faturamento:.2f}")
print(f"3.6 Média diária de reclamações — {media_reclamacoes:.2f}")

i_maior_pedidos = np.argmax(pedidos)
i_menor_pedidos = np.argmin(pedidos)
i_maior_faturamento = np.argmax(faturamento)
i_menor_faturamento = np.argmin(faturamento)

maior_faturamento = faturamento[i_maior_faturamento]
menor_faturamento = faturamento[i_menor_faturamento]
amplitude_faturamento = maior_faturamento - menor_faturamento

print("\nParte 4 — Análise por dia")
print("4.1 Dia com mais pedidos", dias[i_maior_pedidos])
print("4.1 Dia com menos pedidos", dias[i_menor_pedidos])
print("4.2 Dia com maior faturamento", dias[i_maior_faturamento])
print("4.2 Dia com menor faturamento", dias[i_menor_faturamento])
print(f"4.3 Amplitude do faturamento — R$ {amplitude_faturamento:.2f}")

mais_de_150_pedidos = pedidos > 150

print("\nParte 5 — Máscaras e filtros")
print("5.1 Dias com faturamento superior a R$ 12.000", dias[faturamento > 12000])
print("5.2 Dias com mais de 150 pedidos", dias[mais_de_150_pedidos])
print("5.3 Dias com mais de 4 reclamações", dias[reclamacoes > 4])
print("5.4 Faturamento dos dias com mais de 150 pedidos",
      faturamento[mais_de_150_pedidos])
print("5.5 Dias com faturamento abaixo da média",
      np.count_nonzero(faturamento < media_faturamento))

alto_desempenho = faturamento >= 20000
classificacao = np.where(
    alto_desempenho, "Alto desempenho", "Desempenho normal"
)
qtd_alto_desempenho = np.count_nonzero(alto_desempenho)
qtd_desempenho_normal = np.count_nonzero(~alto_desempenho)

print("\nParte 6 — Classificação dos dias")
print("6.1 Dia e classificação")
print(np.column_stack((dias, classificacao)))
print("6.2 Dias de alto desempenho", qtd_alto_desempenho)
print("6.2 Dias de desempenho normal", qtd_desempenho_normal)

print("\nParte 7 — Estatística")
print(f"Média — R$ {media_faturamento:.2f}")
print(f"Mediana — R$ {np.median(faturamento):.2f}")
print(f"Desvio padrão — R$ {np.std(faturamento):.2f}")
print(f"Mínimo — R$ {menor_faturamento:.2f}")
print(f"Máximo — R$ {maior_faturamento:.2f}")

medias = np.mean(dados, axis=0)
maiores = np.maximum.reduce(dados, axis=0)
menores = np.minimum.reduce(dados, axis=0)

variacoes_relativas = (maiores - menores) / medias * 100
i_maior_variacao = np.argmax(variacoes_relativas)

print("\nParte 8 — Trabalhando com axis")
print("Ordem dos indicadores", indicadores)
print("8.1 Médias", np.round(medias, 2))
print("8.2 Maiores valores", maiores)
print("8.3 Menores valores", menores)
print("8.4 Amplitudes relativas (%)", np.round(variacoes_relativas, 2))
print("Indicador com maior variação relativa", indicadores[i_maior_variacao])

i_maior_reclamacoes = np.argmax(reclamacoes)
pedidos_e_faturamento = mais_de_150_pedidos & (faturamento > 20000)

print("\nParte 9 — Análise integrada")
print("9.1 Melhor dia por faturamento", dias[i_maior_faturamento])
print("9.2 Foi também o dia com mais pedidos?",
      np.where(i_maior_faturamento == i_maior_pedidos, "Sim", "Não"))
print("9.3 Foi também o dia com mais reclamações?",
      np.where(i_maior_faturamento == i_maior_reclamacoes, "Sim", "Não"))
print("9.4 Dias com mais de 150 pedidos e faturamento superior a R$ 20.000",
      np.count_nonzero(pedidos_e_faturamento))

print("\nRESUMO DA SEMANA")
print(f"Total de pedidos — {total_pedidos:.0f}")
print(f"Faturamento total — R$ {total_faturamento:.2f}")
print(f"Média diária de pedidos — {media_pedidos:.2f}")
print(f"Média diária de faturamento — R$ {media_faturamento:.2f}")
print("Melhor dia por faturamento", dias[i_maior_faturamento])
print("Pior dia por faturamento", dias[i_menor_faturamento])
print(f"Maior número de reclamações — {reclamacoes[i_maior_reclamacoes]:.0f}")
print("Dias de alto desempenho", dias[alto_desempenho])
```
