```python
import pandas as pd


dados = {
    'Cidade': ['Passo Fundo']*12 + ['Ijuí']*12 + ['Santa Rosa']*12,
    'Mes': list(range(1, 13))*3,
    'Temperatura': [
        24, 23, 22, 19, 16, 13, 12, 15, 17, 19, 21, 23,
        27, 26, 24, 21, 18, 15, 14, 17, 19, 22, 24, 26,
        28, 27, 25, 22, 19, 16, 15, 18, 20, 23, 25, 27
    ],
    'Deficit': [
        30, 25, 15, 10, 5, 0, 0, 5, 10, 15, 20, 25,
        40, 35, 25, 15, 10, 0, 0, 5, 15, 20, 30, 35,
        45, 40, 30, 20, 10, 0, 0, 10, 20, 25, 35, 40
    ]
}

df = pd.DataFrame(dados)


cidade = input("Digite sua cidade (Passo Fundo, Ijuí ou Santa Rosa): ")

df_cidade = df[df['Cidade'] == cidade]

if not df_cidade.empty:


    print("\n1 - TODAS AS INFORMAÇÕES DA CIDADE")
    display(df_cidade)


    print("\n2 - MÊS DE MENOR TEMPERATURA")

    menor = df_cidade.loc[[df_cidade['Temperatura'].idxmin()]]

    display(menor[['Cidade', 'Mes', 'Temperatura']])


    print("\n3 - MÊS DE MAIOR DÉFICIT HÍDRICO")

    maior = df_cidade.loc[[df_cidade['Deficit'].idxmax()]]

    display(maior[['Cidade', 'Mes', 'Deficit']])


    print("\n4 - MÉDIA DE TEMPERATURA DAS ESTAÇÕES")

    estacoes = {
        'Verão': [12, 1, 2],

        'Outono': [3, 4, 5],
        'Inverno': [6, 7, 8],
        'Primavera': [9, 10, 11]
    }

    medias = []

    for estacao, meses in estacoes.items():
        media = df_cidade[
            df_cidade['Mes'].isin(meses)
        ]['Temperatura'].mean()

        medias.append([estacao, round(media, 2)])

    tabela_medias = pd.DataFrame(
        medias,
        columns=['Estação', 'Temperatura Média (°C)']
    )

    display(tabela_medias)

else:
    print("Cidade não encontrada!")
```
